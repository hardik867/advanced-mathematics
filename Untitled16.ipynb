```python
import numpy as np
import pandas as pd
import kagglehub
import os
```

```python
r = 102316005
a = 0.05 * (r % 7)
b = 0.3 * ((r%5) + 1)
print("a = ",a)
print("b = ",b)
```

```python
path = kagglehub.dataset_download(
    "shrutibhargava94/india-air-quality-data"
)
file = os.path.join(path,"data.csv")
df = pd.read_csv(file,encoding = "latin-1")
```

```python
x = df["no2"].dropna()
x = x[x >= 0].values
z = x + a * np.sin(b * x)
mu = np.mean(z)
var = np.var(z)
lam = 1 / (2 * var)
c = np.sqrt(lam / np.pi)
print("mu = ",mu)
print("lambda = ",lam)
print("c = ",c)
```
